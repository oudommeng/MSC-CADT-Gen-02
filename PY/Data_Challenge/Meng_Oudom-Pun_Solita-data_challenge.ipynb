# Data Loading

```python
customers = pd.read_csv('data/Customer.csv')
date = pd.read_csv('data/Date.csv')
products = pd.read_csv('data/Product.csv')
resellers = pd.read_csv('data/Reseller.csv')
sales_order = pd.read_csv('data/Sales_Order.csv')
sales_territory = pd.read_csv('data/Sales_Territory.csv')
sales = pd.read_csv('data/Sales.csv')
```

# Check for Missing Values and Duplicates

```python
dataframes = {
    "Customers": customers,
    "Date": date,
    "Products": products,
    "Resellers": resellers,
    "Sales Order": sales_order,
    "Sales Territory": sales_territory,
    "Sales": sales
}

for name, df in dataframes.items():
    print(f"--- {name} ---")
    print(f"Shape: {df.shape}")
    print(f"Columns: {df.columns.tolist()}")
    print("Missing Values:")
    print(df.isnull().sum()[df.isnull().sum() > 0])
    print(f"Duplicate Rows: {df.duplicated().sum()}\n")
```

# Data Merging

```python
# Merge with Products
master_df = pd.merge(sales, products, on='ProductKey', how='left')

# Merge with Customers
master_df = pd.merge(master_df, customers, on='CustomerKey', how='left')

# Merge with Resellers
master_df = pd.merge(master_df, resellers, on='ResellerKey', how='left')

# Merge with Sales Territory
master_df = pd.merge(master_df, sales_territory, on='SalesTerritoryKey', how='left')

# Merge with Sales Order
master_df = pd.merge(master_df, sales_order, on='SalesOrderLineKey', how='left')

# Merge with Date (on OrderDateKey)
master_df = pd.merge(master_df, date, left_on='OrderDateKey', right_on='DateKey', how='left')

# Inspect the Master DataFrame
print(f"Master DataFrame Shape: {master_df.shape}")
print(f"Sales DataFrame Shape: {sales.shape}") # For verification
master_df.head()
```

# Data Cleaning

```python
# Verify row counts
assert len(master_df) == len(sales), f"Row mismatch! Master: {len(master_df)}, Sales: {len(sales)}"
print("Row count verification passed.")

# Check current types
print(master_df.dtypes)

# Convert 'Date' to datetime
if 'Date' in master_df.columns:
    master_df['Date'] = pd.to_datetime(master_df['Date'])
    print("Converted 'Date' to datetime object.")

# Verify conversion
print(master_df['Date'].dtype)
```

# Part 2: Descriptive Analytics

```python
# Global Financials
total_sales = master_df['Sales Amount'].sum()
total_cost = master_df['Total Product Cost'].sum()
total_profit = total_sales - total_cost

print(f"Total Sales Amount: ${total_sales:,.2f}")
print(f"Total Product Cost: ${total_cost:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
```

```python
# Regional Performance

# Country with highest Sales Amount
country_sales = master_df.groupby('Country')['Sales Amount'].sum().sort_values(ascending=False)
print(f"Highest Sales Country: {country_sales.index[0]} (${country_sales.iloc[0]:,.2f})")

# Region with lowest Sales Amount
region_sales = master_df.groupby('Region')['Sales Amount'].sum().sort_values(ascending=True)
print(f"Lowest Sales Region: {region_sales.index[0]} (${region_sales.iloc[0]:,.2f})")
```

```python
# Product Insights

# Category revenue driver
category_sales = master_df.groupby('Category')['Sales Amount'].sum().sort_values(ascending=False)
print(f"Primary Revenue Driver (Category): {category_sales.index[0]} (${category_sales.iloc[0]:,.2f})")

# Top 5 Selling Products
top_products = master_df.groupby('Product')['Sales Amount'].sum().sort_values(ascending=False).head(5)
print("\nTop 5 Selling Products:")
print(top_products)
```

# Part 3: Temporal Analysis & Trends

```python
# Yearly Trends
yearly_sales = master_df.groupby('Fiscal Year')['Sales Amount'].sum()
print("Sales by Fiscal Year:")
print(yearly_sales)
print(f"Is the company growing? {'Yes' if yearly_sales.is_monotonic_increasing else 'No/Mixed'}")
```

```python
# Monthly Seasonality

# Create Year-Month column for aggregation
master_df['YearMonth'] = master_df['Date'].dt.to_period('M')
monthly_sales = master_df.groupby('YearMonth')['Sales Amount'].sum()

# Highest sales month
best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()
print(f"Highest Sales Month: {best_month} (${best_month_sales:,.2f})")
```

# Visualization

```python
# Revenue Trend Line
plt.figure(figsize=(12, 6))
# Convert period index to string for plotting consistency
monthly_sales.index = monthly_sales.index.astype(str)
monthly_sales.plot(kind='line', marker='o', label='Sales Amount')
plt.title('Revenue Trend Over Time')
plt.xlabel('Month')
plt.ylabel('Sales Amount')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
# Category Breakdown
plt.figure(figsize=(10, 6))
category_sales.plot(kind='bar', color='skyblue', label='Sales Amount')
plt.title('Total Sales by Product Category')
plt.xlabel('Category')
plt.ylabel('Sales Amount')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
# Profit by Country

# Calculate profit by country
country_profit = master_df.groupby('Country')['Sales Amount'].sum() - master_df.groupby('Country')['Total Product Cost'].sum()
country_profit = country_profit.sort_values(ascending=False)

plt.figure(figsize=(10, 6))
country_profit.plot(kind='bar', color='green', label='Total Profit')
plt.title('Total Profit by Country')
plt.xlabel('Country')
plt.ylabel('Total Profit')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

# Part 5: Advanced Challenge

## Sales Channel Analysis

```python
# Calculate Average Order Value by Channel
channel_aov = master_df.groupby('Channel')['Sales Amount'].mean()

print("Average Order Value by Channel:")
print(channel_aov)

# Determine which channel has higher value
best_channel = channel_aov.idxmax()
print(f"\n{best_channel} channel has higher average order value: ${channel_aov[best_channel]:,.2f}")

# Visualize
plt.figure(figsize=(8, 5))
channel_aov.plot(kind='bar', color=['steelblue', 'coral'])
plt.title('Average Order Value by Sales Channel')
plt.xlabel('Channel')
plt.ylabel('Average Order Value ($)')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## Profit Margin Analysis

```python
# Calculate profit margin by Subcategory
subcategory_data = master_df.groupby('Subcategory').agg({
    'Sales Amount': 'sum',
    'Total Product Cost': 'sum'
})

subcategory_data['Total Profit'] = subcategory_data['Sales Amount'] - subcategory_data['Total Product Cost']
subcategory_data['Profit Margin %'] = (subcategory_data['Total Profit'] / subcategory_data['Sales Amount']) * 100

# Sort by profit margin
subcategory_data = subcategory_data.sort_values('Profit Margin %', ascending=False)

print("Top 10 Subcategories by Profit Margin:")
print(subcategory_data[['Sales Amount', 'Total Profit', 'Profit Margin %']].head(10))

most_profitable = subcategory_data.index[0]
print(f"\nMost profitable Subcategory: {most_profitable} ({subcategory_data['Profit Margin %'].iloc[0]:.2f}%)")

# Visualize top 10
plt.figure(figsize=(12, 6))
subcategory_data['Profit Margin %'].head(10).plot(kind='bar', color='green')
plt.title('Top 10 Subcategories by Profit Margin %')
plt.xlabel('Subcategory')
plt.ylabel('Profit Margin %')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Box Plot: Sales Distribution by Month
import matplotlib.pyplot as plt

# Extract month name from Date column
master_df['Month_Name'] = master_df['Date'].dt.month_name()

# Create ordered month list for proper sorting
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']

# Prepare data for box plot
sales_by_month = [master_df[master_df['Month_Name'] == month]['Sales Amount'].values 
                  for month in month_order if month in master_df['Month_Name'].unique()]

# Get actual months present in data
months_present = [month for month in month_order if month in master_df['Month_Name'].unique()]

# Create box plot
plt.figure(figsize=(14, 6))
box = plt.boxplot(sales_by_month, labels=months_present, patch_artist=True)

# Customize colors
for patch in box['boxes']:
    patch.set_facecolor('lightblue')

plt.title('Sales Distribution by Month (Box Plot)', fontsize=14, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Sales Amount ($)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Print summary statistics by month
print("\nSales Summary Statistics by Month:")
monthly_stats = master_df.groupby('Month_Name')['Sales Amount'].describe()[['mean', '50%', 'std']]
monthly_stats.columns = ['Mean', 'Median', 'Std Dev']
monthly_stats = monthly_stats.reindex(months_present)
print(monthly_stats)
```

## 🔥 Sales Heatmap (Month vs Year)

```python
import seaborn as sns

# Create pivot for heatmap
master_df['Year'] = master_df['Date'].dt.year
master_df['Month'] = master_df['Date'].dt.month
pivot_sales = master_df.pivot_table(values='Sales Amount', 
                                      index='Month', 
                                      columns='Year', 
                                      aggfunc='sum')

plt.figure(figsize=(12, 8))
sns.heatmap(pivot_sales/1000000, annot=True, fmt='.2f', cmap='YlGnBu', 
            cbar_kws={'label': 'Sales Amount (Millions $)'})
plt.title('Sales Heatmap: Month vs Year (in Millions)', fontsize=14, fontweight='bold')
plt.ylabel('Month')
plt.xlabel('Year')
plt.tight_layout()
plt.show()
```

## 🏆 Top 10 Products (Horizontal Bar)

## 👥 Customer Segmentation Analysis

```python
# Customer Analysis (Internet sales if exist)
if 'Internet' in master_df['Channel'].unique():
    internet_customers = master_df[master_df['Channel'] == 'Internet']
    customer_value = internet_customers.groupby('CustomerKey').agg({
        'Sales Amount': ['sum', 'count', 'mean']
    }).round(2)
    customer_value.columns = ['Total_Spent', 'Order_Count', 'Avg_Order_Value']
    customer_value = customer_value.sort_values('Total_Spent', ascending=False)
    
    print("Top 10 Internet Customers by Total Spending:")
    print(customer_value.head(10))
    
    # Visualize customer distribution
    plt.figure(figsize=(12, 5))
    
    plt.subplot(1, 2, 1)
    customer_value['Total_Spent'].head(20).plot(kind='bar', color='coral')
    plt.title('Top 20 Customers by Total Spending')
    plt.ylabel('Total Spent ($)')
    plt.xlabel('Customer')
    plt.xticks([])
    
    plt.subplot(1, 2, 2)
    customer_value['Order_Count'].head(20).plot(kind='bar', color='steelblue')
    plt.title('Top 20 Customers by Order Count')
    plt.ylabel('Number of Orders')
    plt.xlabel('Customer')
    plt.xticks([])
    
    plt.tight_layout()
    plt.show()
else:
    print("Note: All sales are Reseller channel. Customer-level analysis not applicable.")
```

## 📈 Year-over-Year Growth Analysis

```python
# YoY Growth Rate
yearly_sales_sorted = yearly_sales.sort_index()
growth_rate = yearly_sales_sorted.pct_change() * 100

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
yearly_sales_sorted.plot(kind='bar', color='skyblue')
plt.title('Sales by Fiscal Year', fontweight='bold')
plt.ylabel('Sales Amount ($)')
plt.xlabel('Fiscal Year')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.subplot(1, 2, 2)
growth_rate.dropna().plot(kind='bar', color='green')
plt.title('Year-over-Year Growth Rate', fontweight='bold')
plt.ylabel('Growth Rate (%)')
plt.xlabel('Fiscal Year')
plt.xticks(rotation=45)
plt.axhline(y=0, color='red', linestyle='--', linewidth=1)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\nYear-over-Year Growth Rates:")
for year, rate in growth_rate.dropna().items():
    print(f"{year}: {rate:+.2f}%")
```

## 📅 Quarterly Performance Trends

```python
# Quarterly trends
master_df['Quarter'] = master_df['Date'].dt.to_period('Q')
quarterly_sales = master_df.groupby('Quarter')['Sales Amount'].sum()
quarterly_profit = master_df.groupby('Quarter').apply(lambda x: x['Sales Amount'].sum() - x['Total Product Cost'].sum())

plt.figure(figsize=(14, 6))
quarterly_sales.index = quarterly_sales.index.astype(str)
quarterly_profit.index = quarterly_profit.index.astype(str)

x = range(len(quarterly_sales))
plt.plot(x, quarterly_sales.values, marker='o', linewidth=2, markersize=8, label='Sales Amount', color='blue')
plt.plot(x, quarterly_profit.values, marker='s', linewidth=2, markersize=8, label='Profit', color='green')

plt.title('Quarterly Sales & Profit Trend', fontsize=14, fontweight='bold')
plt.xlabel('Quarter')
plt.ylabel('Amount ($)')
plt.xticks(x, quarterly_sales.index, rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Show best and worst quarters
best_quarter = quarterly_sales.idxmax()
worst_quarter = quarterly_sales.idxmin()
print(f"\n📈 Best Quarter: {best_quarter} (${quarterly_sales.max():,.2f})")
print(f"📉 Worst Quarter: {worst_quarter} (${quarterly_sales.min():,.2f})")
```

## 🔗 Correlation Analysis

```python
# Correlation between numeric variables
numeric_cols = ['Order Quantity', 'Unit Price', 'Sales Amount', 'Total Product Cost', 'Unit Price Discount Pct']
correlation = master_df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Matrix - Sales Variables', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n📊 Key Insights:")
print(f"Sales Amount vs Order Quantity correlation: {correlation.loc['Sales Amount', 'Order Quantity']:.3f}")
print(f"Sales Amount vs Unit Price correlation: {correlation.loc['Sales Amount', 'Unit Price']:.3f}")
print(f"Unit Price vs Discount correlation: {correlation.loc['Unit Price', 'Unit Price Discount Pct']:.3f}")
```
