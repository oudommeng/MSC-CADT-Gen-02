# Random and Checking missing value 

```python
data = pd.DataFrame(np.random.randn(10, 5), index= None, columns=['id', 'name', 'age', 'emial', 'tel'])
# print(data)
df = data.reindex(['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j'])
print(df)
```

```python
df['id'].fillna(-1)
```

```python
# Check null 
chk = df['id'].isnull()
print(chk)
```
