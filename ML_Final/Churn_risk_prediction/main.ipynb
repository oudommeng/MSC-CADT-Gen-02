# Load Data

This section handles loading the dataset into a pandas DataFrame. It automatically detects the active environment and loads the data accordingly:
- **Google Colab**: Mounts Google Drive and loads the dataset from the specific Colab path.
- **Local Environment**: Loads the dataset from a relative local path (`dataset/customer_churn_dataset.csv`).


```python
try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    from google.colab import drive
    drive.mount("/content/drive")
    file_path = '/content/drive/MyDrive/MSC/ML/Dataset/customer_churn_dataset.csv'
    print("Running in Colab: Using Google Drive path.")
else:
    file_path = 'dataset/customer_churn_dataset.csv'
    print("Running Locally: Using local relative path.")
    
if os.path.exists(file_path):
    df = pd.read_csv(file_path)
    print("✅ Successfully loaded the dataset!")
    display(df.head())
else:
    print(f"❌ Error: Could not find the dataset at: {file_path}")
```
